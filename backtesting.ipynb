```python
# Method for imputing stock data from dataframe depending on stock type
def imputeData(dataframe:DataFrame) -> DataFrame:
    knn_imputer = joblib.load("./model/imputers/knn_imputer.joblib")
    median_imputer = joblib.load("./model/imputers/median_imputer.joblib")
    ffil_median_imputer = joblib.load("./model/imputers/ffil_median_imputer.joblib")
    # print(dataframe.columns)
    
    dataframe[knn_impute_features] = knn_imputer.transform(dataframe[knn_impute_features])
    dataframe[median_impute_features] = median_imputer.transform(dataframe[median_impute_features])
    dataframe[forward_fill_features] = ffil_median_imputer.transform(dataframe[forward_fill_features])
    # dataframe[forward_fill_features] = dataframe[forward_fill_features].ffill().bfill().interpolate()
    # for col in forward_fill_features:
    #     if dataframe[col].isna().all():
    #         ffil_median_imputer.transform(dataframe[forward_fill_features])
            
    return dataframe
```

# Utilities Methods

```python
def closest_elements(nums, target, numOfElements: int):
    nums_sorted = sorted(nums, key=lambda x: abs(x[1] - target))
    return nums_sorted[:numOfElements]
```

```python
loaded_rf = joblib.load("./model/random_forest.joblib")
def predict_model(pre_df: DataFrame):
    pre_df_aligned = pre_df.reindex(columns=expected_features)

    pre_df_imputed = imputeData(pre_df_aligned)

    imputed_df = pd.DataFrame(pre_df_imputed, columns=expected_features)
    
    Y_pred = loaded_rf.predict(imputed_df)
    pred_df = pd.DataFrame(Y_pred, columns=["Y_pred"])
    new_df = pd.concat([pred_df.reset_index(drop=True), imputed_df.reset_index(drop=True)], axis=1)

    return new_df
```

```python
def analysisOnCorr(df, selectedValue: str="Y", showAll: bool=False):
  corr = df.corr()
  keys = corr.columns
  values = corr[selectedValue]
  hm = {}
  for i in range(len(values)):
    hm[keys[i]] = values[i]

  tuplesLi = sorted(hm.items(), key=lambda x: x[1], reverse=True)
  # print(tuplesLi)

  temp = []
  if showAll:
    for i in range(len(tuplesLi)):
      temp.append(tuplesLi[i][0])
      # print(tuplesLi[i][0], tuplesLi[i][1])
  else:
    print("Top Postive Coorelation")
    for i in range(7):
      temp.append(tuplesLi[i][0])
      # print(tuplesLi[i][0], tuplesLi[i][1])

    print("Top Negative Coorelation")
    for j in range(len(tuplesLi)-1, len(tuplesLi)-7, -1):
      temp.append(tuplesLi[j][0])
      # print(tuplesLi[j][0], tuplesLi[j][1])
    
  temp_df = df[temp]

  return tuplesLi
```

```python

def selectedClosestElement(df:DataFrame, numOfElements: int):
    ewma_y = pd.Series(df["Y_pred"]).ewm(span=20, adjust=False).mean()
    ewma_min = ewma_y.min()
    ewma_max = ewma_y.max()
    ewma_normalized = 2 * (ewma_y - ewma_min) / (ewma_max - ewma_min) - 1
    # print(df["Y_pred"], ewma_normalized)
    investableLi = [item for item in YcoorelationDecreasingOrderLi if item[0] in investable_assets]
    selected_Elements = closest_elements(investableLi, ewma_normalized.iat[-1], numOfElements)
    return selected_Elements
```
